```python
import pandas as pd


dados_titulos_steam = {
    'ID':[],
    'Nome do Jogo':[],
    'Data de Lançamento Steam':[],
    'Preço do Jogo':[]
}
dados_promocao = {
    'ID_jogo':[],
    'Data da Promoção':[],
    'Porcentagem da Promoção':[]
    }

def inserir_dados_jogo():
  ID = len(dados_titulos_steam['ID'])+1
  nome_jogo = input('Nome do Jogo: ')
  data_lancamento = input('Data de Lançamento Steam: ')
  preco_jogo = input('Preço do Jogo: ')
  dados_titulos_steam['Nome do Jogo'].append(nome_jogo)
  dados_titulos_steam['Data de Lançamento Steam'].append(data_lancamento)
  dados_titulos_steam['Preço do Jogo'].append(preco_jogo)
  dados_titulos_steam['ID'].append(ID)

def inserir_dados_promocao():
  ID_jogo = input('ID_Jogo: ')
  data_promocao = input('Data da Promoção: ')
  porcentagem_promocao = input('Porcentagem da Promoção: ')
  dados_promocao['ID_jogo'].append(ID_jogo)
  dados_promocao['Data da Promoção'].append(data_promocao)
  dados_promocao['Porcentagem da Promoção'].append(porcentagem_promocao)
```
